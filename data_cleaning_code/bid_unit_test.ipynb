```python
import pandas as pd
import numpy as np
from datetime import timedelta
from nemosis import static_table, dynamic_data_compiler
import plotly.express as px
```

```python
raw_data_cache = '/Volumes/T7/NEMO_data'
```

```python
# Download the latest Generators and Scheduled Loads table. The 
# update_static_file=True argument forces nemosis to download a new copy of 
# file from AEMO even if a copy already exists in the cache.
dispatch_units = static_table(table_name='Generators and Scheduled Loads', 
                              raw_data_location=raw_data_cache,
                              update_static_file=False)
dispatch_units.head()
```

```python
# Use pandas filtering to find the DUID of Daydream solar farm
dispatch_units[dispatch_units['Station Name'].str.contains('Murray Power')]
```

```python
# Get the volume bids for the energy market.
volume_bids = dynamic_data_compiler(start_time='2021/03/01 00:00:00',
                                   end_time='2021/04/10 00:00:00',
                                   table_name='BIDPEROFFER_D',
                                   raw_data_location=raw_data_cache)
volume_bids = volume_bids[(volume_bids['DUID'] == 'MURRAY') & 
                          (volume_bids['BIDTYPE'] == 'ENERGY')]
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
display(volume_bids)
```

```python
# Reshape the data such that each bid volume is on its own row.
volume_bids = pd.melt(volume_bids, id_vars=['INTERVAL_DATETIME'], 
                      value_vars=['BANDAVAIL1', 'BANDAVAIL2', 'BANDAVAIL3', 'BANDAVAIL4', 
                                  'BANDAVAIL5', 'BANDAVAIL6', 'BANDAVAIL7', 'BANDAVAIL8', 
                                  'BANDAVAIL9', 'BANDAVAIL10'],
                      var_name='BIDBAND', value_name='BIDVOLUME')
volume_bids.head()
```

```python
# Get the price bids.
price_bids = dynamic_data_compiler(start_time='2021/03/01 00:00:00',
                                   end_time='2021/04/10 00:00:00',
                                   table_name='BIDDAYOFFER_D',
                                   raw_data_location=raw_data_cache)
price_bids = price_bids[price_bids['DUID'] == 'MURRAY']
price_bids.head()
```

```python
# Reshape the data such that each bid volume is on its own row.
price_bids = pd.melt(price_bids, id_vars=['SETTLEMENTDATE'], 
                     value_vars=['PRICEBAND1', 'PRICEBAND2', 'PRICEBAND3', 'PRICEBAND4', 
                                 'PRICEBAND5', 'PRICEBAND6', 'PRICEBAND7', 'PRICEBAND8', 
                                 'PRICEBAND9', 'PRICEBAND10'],
                     var_name='BIDBAND', value_name='BIDPRICE')
price_bids.head()
```

```python
# Price bids are submitted daily and are applicable from the beginning of the market day. 
# The first interval of the market day is 04:05:00, so we will add an applicable from column
# to make it easier to match volume and price bids.
price_bids['APPLICABLEFROM'] = price_bids['SETTLEMENTDATE'] + timedelta(hours=4, minutes=5)
price_bids.head()
```

```python
# To make it easier to match volume and price bids we will also convert the BIDBAND values to numeric
price_bids['BIDBAND'] = pd.to_numeric(price_bids['BIDBAND'].str[9:])
price_bids.head()
```

```python
volume_bids['BIDBAND'] = pd.to_numeric(volume_bids['BIDBAND'].str[9:])
volume_bids.head()
```

```python
# Now we match each volume bid with the most recent price bid.
bids = pd.merge_asof(volume_bids.sort_values('INTERVAL_DATETIME'), 
                     price_bids.sort_values('APPLICABLEFROM'), 
                     left_on='INTERVAL_DATETIME', right_on='APPLICABLEFROM',
                     by='BIDBAND')
bids
```

```python
# Filter from midnight April 7 up to (but not including) midnight April 8
bids_single_day = bids[
    (bids['INTERVAL_DATETIME'] >= "2021-04-07 00:00:00") &
    (bids['INTERVAL_DATETIME'] <  "2021-04-08 00:00:00")
].copy()

# Plot
fig = px.area(
    bids_single_day.sort_values('BIDBAND'),
    x='INTERVAL_DATETIME',
    y='BIDVOLUME',
    color='BIDPRICE'
)
fig.update_layout(yaxis_title="MW")
fig.show()
```

```python
import plotly.graph_objects as go
import datetime

# 1) Filter for 2021-04-07 from 18:00 to just before 18:05
start_dt = pd.to_datetime("2021-04-07 18:00:00")
end_dt   = pd.to_datetime("2021-04-07 18:05:00")

bids_6pm = bids[
    (bids['INTERVAL_DATETIME'] >= start_dt) &
    (bids['INTERVAL_DATETIME'] <  end_dt)
].copy()

# 2) Sort by ascending bid price
bids_6pm.sort_values('BIDPRICE', inplace=True)

# 3) Create a cumulative volume column to represent the supply curve “x-axis”
bids_6pm['CUMVOL'] = bids_6pm['BIDVOLUME'].cumsum()

# 4) Step-plot of Price vs. Cumulative Volume
fig = go.Figure()
fig.add_trace(
    go.Scatter(
        x=bids_6pm['CUMVOL'],
        y=bids_6pm['BIDPRICE'],
        mode='lines+markers',
        line_shape='hv',  # stepped lines (horizontal then vertical)
        name='6pm Supply Curve'
    )
)

fig.update_layout(
    title='Supply Bid Curve at 6:00–6:05 PM on 2021-04-07',
    xaxis_title='Cumulative Offered MW',
    yaxis_title='Price ($/MWh)'
)
fig.show()
```

```python
import datetime
import plotly.graph_objects as go

# Create a date range for the month
dates = pd.date_range(start='2021-04-01', end='2021-04-30')

for single_day in dates:
    # Convert to a date string, e.g. "2021-04-07"
    day_str = single_day.strftime('%Y-%m-%d')
    
    # 1) Filter for the single_day between 18:00 and 18:05
    start_dt = pd.to_datetime(day_str + " 18:00:00")
    end_dt   = pd.to_datetime(day_str + " 18:05:00")
    
    bids_6pm = bids[
        (bids['INTERVAL_DATETIME'] >= start_dt) &
        (bids['INTERVAL_DATETIME'] <  end_dt)
    ].copy()
    
    # If no rows, skip
    if bids_6pm.empty:
        print(f"No 6pm data on {day_str}, skipping.")
        continue
    
    # 2) Sort by ascending bid price
    bids_6pm.sort_values('BIDPRICE', inplace=True)
    
    # 3) Cumulative volume for the supply curve
    bids_6pm['CUMVOL'] = bids_6pm['BIDVOLUME'].cumsum()
    
    # 4) Step-plot
    fig = go.Figure()
    fig.add_trace(
        go.Scatter(
            x=bids_6pm['CUMVOL'],
            y=bids_6pm['BIDPRICE'],
            mode='lines+markers',
            line_shape='hv',
            name='6pm Supply Curve'
        )
    )

    fig.update_layout(
        title=f"Supply Bid Curve at 6:00–6:05 PM on {day_str}",
        xaxis_title='Cumulative Offered MW',
        yaxis_title='Price ($/MWh)'
    )
    
    # Show or save the figure
    fig.show()
    # Optionally save to file: fig.write_html(f"supply_curve_{day_str}.html")
```
