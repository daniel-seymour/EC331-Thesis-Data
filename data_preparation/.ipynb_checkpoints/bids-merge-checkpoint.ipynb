```python
import os
import glob
import pandas as pd
import re

# List of band columns we want to melt
band_cols = [f"BANDAVAIL{i}" for i in range(1, 11)]

def melt_one_file(input_path, output_dir):
    """
    Reads one Parquet file into memory, melts the BANDAVAIL columns,
    and saves the melted data to a new Parquet file in output_dir.
    """
    try:
        # 1) Read the single Parquet file into a pandas DataFrame
        df = pd.read_parquet(input_path)

        # 2) Identify all other columns as id_vars
        id_vars_cols = [col for col in df.columns if col not in band_cols]

        # 3) Melt the DataFrame
        df_long = pd.melt(
            df,
            id_vars=id_vars_cols,
            value_vars=band_cols,
            var_name="BIDBAND",
            value_name="BIDVOLUME"
        )

        # 4) Convert "BIDBAND" from e.g. "BANDAVAIL1" -> integer 1
        df_long["BIDBAND"] = df_long["BIDBAND"].str.extract(r"BANDAVAIL(\d+)").astype(int)

        # 5) Construct the output file path - FIXED
        base_name = os.path.basename(input_path)
        
        # Get the file number from the original filename
        match = re.search(r'part\.(\d+)\.parquet', base_name)
        if match:
            file_num = match.group(1)
            # Create new filename with consistent formatting
            melted_name = f"part.{file_num}_melted.parquet"
        else:
            # If no match found, use a safe fallback
            melted_name = base_name.replace(".parquet", "_melted.parquet")
            
        output_path = os.path.join(output_dir, melted_name)

        # 6) Save melted data to a new Parquet file
        df_long.to_parquet(output_path, index=False)
        print(f"Melted file saved to {output_path}")
        return output_path
    except Exception as e:
        print(f"Error processing {input_path}: {str(e)}")
        return None


def main():
    input_dir = "/Volumes/T7/bid-volume-enriched_sa1-A4"  # Where your original Parquet files live
    output_dir = "/Volumes/T7/bid-volume-melted-files"    # Where to save melted files
    os.makedirs(output_dir, exist_ok=True)

    # Gather Parquet files
    parquet_files = glob.glob(os.path.join(input_dir, "*.parquet"))
    
    # Extract numerical part for proper sorting
    def get_file_number(filepath):
        match = re.search(r'part\.(\d+)\.parquet', os.path.basename(filepath))
        if match:
            return int(match.group(1))
        return 0  # Default for files that don't match the pattern
    
    # Sort files numerically by their part number
    parquet_files = sorted(parquet_files, key=get_file_number)
    
    # Exclude hidden macOS files
    parquet_files = [f for f in parquet_files if not os.path.basename(f).startswith("._")]

    print(f"Found {len(parquet_files)} Parquet files in {input_dir}.")

    # Process each file individually and collect results
    processed_files = []
    for i, file_path in enumerate(parquet_files):
        output_path = melt_one_file(file_path, output_dir)
        if output_path:
            processed_files.append(output_path)
        
        # Print progress every 10 files
        if (i + 1) % 10 == 0:
            print(f"Processed {i + 1}/{len(parquet_files)} files")

    print(f"Successfully processed {len(processed_files)} out of {len(parquet_files)} files.")
    
    # Verify output files
    print("\nVerifying output files:")
    for i, file_path in enumerate(processed_files[:10]):  # Print first 10 for verification
        print(f"  {i+1}. {os.path.basename(file_path)}")
    
    print("All files melted and saved!")


if __name__ == "__main__":
    main()
```

```python
import os
import glob
import pandas as pd
from datetime import timedelta

# Define which columns to melt
price_band_cols = [f"PRICEBAND{i}" for i in range(1, 11)]

def melt_one_price_file(input_path, output_dir):
    """
    Reads one Parquet file (price bids) into memory,
    melts the PRICEBAND columns into a single column,
    and saves the melted data to a new Parquet file in output_dir.
    """
    # 1) Read the single Parquet file into a pandas DataFrame
    df = pd.read_parquet(input_path)

    # 2) Identify all other columns as id_vars
    id_vars_cols = [col for col in df.columns if col not in price_band_cols]

    # 3) Melt the DataFrame
    df_long = pd.melt(
        df,
        id_vars=id_vars_cols,
        value_vars=price_band_cols,
        var_name="BIDBAND",
        value_name="BIDPRICE"
    )

    # 4) Convert "BIDBAND" from e.g. "PRICEBAND1" -> integer 1
    df_long["BIDBAND"] = df_long["BIDBAND"].str.extract(r"(\d+)").astype(int)

    # 5) (Optional) Convert other columns to consistent dtypes
    #    Adjust column names if they differ in your data
    if "BIDTYPE" in df_long.columns:
        df_long["BIDTYPE"] = df_long["BIDTYPE"].astype(str)
    if "DUID" in df_long.columns:
        df_long["DUID"] = df_long["DUID"].astype(str)
    if "SETTLEMENTDATE" in df_long.columns:
        df_long["SETTLEMENTDATE"] = pd.to_datetime(df_long["SETTLEMENTDATE"])
        # Add 4 hours, 5 minutes offset for APPLICABLEFROM
        df_long["APPLICABLEFROM"] = df_long["SETTLEMENTDATE"] + timedelta(hours=4, minutes=5)
        df_long["APPLICABLEFROM"] = pd.to_datetime(df_long["APPLICABLEFROM"])

    # 6) Construct the output file path
    base_name = os.path.basename(input_path)
    melted_name = base_name.replace(".parquet", "_melted.parquet")
    output_path = os.path.join(output_dir, melted_name)

    # 7) Save melted data to a new Parquet file
    df_long.to_parquet(output_path, index=False)
    print(f"Melted file saved to {output_path}")

def main():
    input_dir = "/Volumes/T7/bid-price-data-excluding-energy-B2"  # Where your price Parquet files live
    output_dir = "/Volumes/T7/bid-price-melted-files"  # Where to save melted files
    os.makedirs(output_dir, exist_ok=True)

    # Gather Parquet files
    parquet_files = sorted(glob.glob(os.path.join(input_dir, "*.parquet")))
    # Exclude hidden macOS files
    parquet_files = [f for f in parquet_files if not os.path.basename(f).startswith("._")]

    print(f"Found {len(parquet_files)} Parquet files in {input_dir}.")

    # Process each file individually
    for file_path in parquet_files:
        melt_one_price_file(file_path, output_dir)

    print("All price files melted and saved!")

if __name__ == "__main__":
    main()
```

```python
import glob
import os
import dask.dataframe as dd
import pandas as pd
import tempfile

def load_valid_parquet_files(folder_pattern):
    """
    Load all valid parquet files matching the given pattern into a Dask DataFrame.
    Excludes hidden files (like macOS ._* files).
    
    Args:
        folder_pattern: Glob pattern to match parquet files
        
    Returns:
        Dask DataFrame containing the data from all valid parquet files
    """
    # Get all files matching the pattern
    all_files = glob.glob(folder_pattern)
    
    # Filter out hidden files (like macOS ._ files)
    valid_files = [f for f in all_files if not os.path.basename(f).startswith('._')]
    
    # Sort files to ensure consistent reading order
    valid_files = sorted(valid_files)
    
    print(f"Found {len(valid_files)} valid parquet files from pattern: {folder_pattern}")
    
    if not valid_files:
        raise ValueError(f"No valid parquet files found for pattern: {folder_pattern}")
    
    # Load files into a Dask DataFrame
    return dd.read_parquet(valid_files, engine='pyarrow')

def process_data_in_chunks(volume_ddf, price_ddf, output_dir, chunk_size=1000000):
    """
    Process the data chunk by chunk to avoid memory issues.
    
    This method:
    1. Writes each dataframe to temporary CSV files
    2. Loads chunks of data from those files
    3. Merges the chunks
    4. Writes the results to parquet files
    """
    os.makedirs(output_dir, exist_ok=True)
    
    # Write volume dataframe to temporary CSV files
    print("Writing volume data to temporary CSV files...")
    with tempfile.TemporaryDirectory() as volume_temp_dir:
        volume_temp_path = os.path.join(volume_temp_dir, "volume_data_*.csv")
        volume_ddf.to_csv(volume_temp_path, index=False)
        
        # Get the actual CSV files created
        volume_csv_files = sorted(glob.glob(os.path.join(volume_temp_dir, "*.csv")))
        print(f"Volume data written to {len(volume_csv_files)} temporary CSV files")
        
        # Write price dataframe to temporary CSV files
        print("Writing price data to temporary CSV files...")
        with tempfile.TemporaryDirectory() as price_temp_dir:
            price_temp_path = os.path.join(price_temp_dir, "price_data_*.csv")
            price_ddf.to_csv(price_temp_path, index=False)
            
            # Get the actual CSV files created
            price_csv_files = sorted(glob.glob(os.path.join(price_temp_dir, "*.csv")))
            print(f"Price data written to {len(price_csv_files)} temporary CSV files")
            
            # Process each volume CSV file one by one
            for i, volume_file in enumerate(volume_csv_files):
                print(f"Processing volume file {i+1}/{len(volume_csv_files)}: {volume_file}")
                
                # Read the volume data as pandas DataFrame
                volume_df = pd.read_csv(volume_file)
                volume_df["SETTLEMENTDATE"] = pd.to_datetime(volume_df["SETTLEMENTDATE"], errors="coerce")
                
                # Process each price CSV file against this volume file
                for j, price_file in enumerate(price_csv_files):
                    print(f"  Merging with price file {j+1}/{len(price_csv_files)}: {price_file}")
                    
                    # Read the price data as pandas DataFrame
                    price_df = pd.read_csv(price_file)
                    price_df["SETTLEMENTDATE"] = pd.to_datetime(price_df["SETTLEMENTDATE"], errors="coerce")
                    
                    # Merge the DataFrames
                    merged_df = volume_df.merge(
                        price_df,
                        on=["SETTLEMENTDATE", "DUID", "BIDTYPE", "BIDBAND"],
                        how="inner",
                        suffixes=('_volume', '_price')
                    )
                    
                    # If we have merged data, write it to a parquet file
                    if not merged_df.empty:
                        output_file = os.path.join(output_dir, f"merged_v{i}_p{j}.parquet")
                        merged_df.to_parquet(output_file, engine='pyarrow', index=False)
                        print(f"    Wrote {len(merged_df)} rows to {output_file}")
                    else:
                        print(f"    No matching data between these files")

def main():
    # Set up paths for input files
    volume_pattern = "/Volumes/T7/bid-volume-melted-files-A4/*.parquet"
    price_pattern = "/Volumes/T7/bid-price-melted-files-B3/*.parquet"
    
    print("Loading volume data...")
    volume_ddf = load_valid_parquet_files(volume_pattern)
    print(f"Volume data columns: {list(volume_ddf.columns)}")
    print(f"Volume data partitions: {volume_ddf.npartitions}")
    
    print("Loading price data...")
    price_ddf = load_valid_parquet_files(price_pattern)
    print(f"Price data columns: {list(price_ddf.columns)}")
    print(f"Price data partitions: {price_ddf.npartitions}")
    
    # Convert SETTLEMENTDATE to datetime if needed (just defining the operation, not computing)
    print("Defining datetime conversion for SETTLEMENTDATE...")
    volume_ddf = volume_ddf.assign(
        SETTLEMENTDATE=dd.to_datetime(volume_ddf["SETTLEMENTDATE"], errors="coerce")
    )
    price_ddf = price_ddf.assign(
        SETTLEMENTDATE=dd.to_datetime(price_ddf["SETTLEMENTDATE"], errors="coerce")
    )
    
    # Create output directory if it doesn't exist
    output_dir = "/Volumes/T7/bid-merged-fcas"
    print(f"Will save merged data to {output_dir}...")
    
    # Process data in chunks
    process_data_in_chunks(volume_ddf, price_ddf, output_dir)
    
    # Report completion
    print("All done!")
    print(f"Output saved to: {output_dir}")

if __name__ == "__main__":
    main()
```
