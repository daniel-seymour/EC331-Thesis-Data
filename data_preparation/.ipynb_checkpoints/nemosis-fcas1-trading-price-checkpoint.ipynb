```python
from nemosis import cache_compiler
from nemosis import defaults
from nemosis import dynamic_data_compiler
```

```python
# Check and add RAISE1SECRRP and RAISE1SECROP columns if they're not already included
if 'RAISE1SECRRP' not in defaults.table_columns['DISPATCHPRICE']:
    defaults.table_columns['TRADINGPRICE'] += ['RAISE1SECRRP']
if 'RAISE1SECROP' not in defaults.table_columns['DISPATCHPRICE']:
    defaults.table_columns['TRADINGPRICE'] += ['RAISE1SECROP']

start_time = '2023/12/01 00:00:00'  # Set your desired start time
end_time = '2023/12/02 00:00:00'    # Set your desired end time
table = 'TRADINGPRICE'
raw_data_cache = '/Volumes/T7/FCAS_RAISE1SEC-Volume-Bids'

# First, create the cache with feather files
cache_compiler(start_time, end_time, table, raw_data_cache, fformat='feather', rebuild=True)

# If you also want to access the data, use dynamic_data_compiler
# Now the data should be in the cache, so this should be fast
price_data = dynamic_data_compiler(
    start_time, 
    end_time, 
    table, 
    raw_data_cache
)

# Display the columns to verify RAISE1SECRRP and RAISE1SECROP are included
print("Columns in the data:", price_data.columns.tolist())

price_data.head
```
