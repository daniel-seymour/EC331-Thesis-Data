```python
# Ensure 'SETTLEMENTDATE' is in datetime format
dispatch_load_df['SETTLEMENTDATE'] = pd.to_datetime(dispatch_load_df['SETTLEMENTDATE'])

# Filter the DataFrame where 'RAISE1SEC' is not equal to 0
filtered_df = dispatch_load_df[dispatch_load_df['RAISE1SEC'] != 0]

# Extract date from 'SETTLEMENTDATE'
filtered_df['DATE'] = filtered_df['SETTLEMENTDATE'].dt.date

# Count unique DUIDs for each date
firms_per_day = filtered_df.groupby('DATE')['DUID'].nunique().reset_index()

# Rename columns for clarity
firms_per_day.columns = ['DATE', 'NUM_FIRMS']
```

```python
# Ensure 'SETTLEMENTDATE' is in datetime format
dispatch_load_df['SETTLEMENTDATE'] = pd.to_datetime(dispatch_load_df['SETTLEMENTDATE'])

# Filter where 'RAISE1SEC' is not zero
filtered_df = dispatch_load_df[dispatch_load_df['RAISE1SEC'] != 0]

# Extract date from 'SETTLEMENTDATE'
filtered_df['DATE'] = filtered_df['SETTLEMENTDATE'].dt.date

# Total RAISE1SEC dispatched per day
raise1sec_per_day = filtered_df.groupby('DATE')['RAISE1SEC'].sum().reset_index()
raise1sec_per_day.columns = ['DATE', 'TOTAL_RAISE1SEC']

# Total RAISE1SEC dispatched per DUID
raise1sec_per_duid = filtered_df.groupby('DUID')['RAISE1SEC'].sum().reset_index()
raise1sec_per_duid = raise1sec_per_duid.sort_values(by='RAISE1SEC', ascending=False)
raise1sec_per_duid.columns = ['DUID', 'TOTAL_RAISE1SEC']
```

```python
raise1sec_per_duid
```

```python
# Calculate total RAISE1SEC dispatched across all DUIDs
total_raise1sec = raise1sec_per_duid['TOTAL_RAISE1SEC'].sum()

# Calculate percentage contribution for each DUID
raise1sec_per_duid['PERCENTAGE'] = (raise1sec_per_duid['TOTAL_RAISE1SEC'] / total_raise1sec) * 100

raise1sec_per_duid
```
