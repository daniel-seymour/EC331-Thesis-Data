```python
import awswrangler as wr
import time
import pandas as pd

def test_individual_file_reading():
    price_path = "s3://thesis--ec331-s3/melted-price-bids/RAISE1SEC_FILTERED_202310010000/"
    
    print(f"Starting to inspect files in folder: {price_path}")
    
    # List the objects in the folder
    start_time = time.time()
    try:
        all_objects = wr.s3.list_objects(path=price_path)
        # Filter to only include parquet files
        parquet_objects = [obj for obj in all_objects if obj.endswith('.parquet')]
        
        num_files = len(parquet_objects)
        num_ignored = len(all_objects) - num_files
        print(f"Found {num_files} parquet files in {time.time() - start_time:.2f} seconds")
        if num_ignored > 0:
            print(f"Ignored {num_ignored} non-parquet files")
        
        # Try to read each file individually
        successful_dfs = []
        
        for i, file_path in enumerate(parquet_objects):
            file_name = file_path.split('/')[-1]
            print(f"\nAttempting to read file {i+1}/{num_files}: {file_name}")
            
            try:
                file_start_time = time.time()
                file_df = wr.s3.read_parquet(path=file_path)
                file_duration = time.time() - file_start_time
                
                rows = len(file_df)
                cols = len(file_df.columns)
                
                print(f"✓ Successfully read in {file_duration:.2f} seconds")
                print(f"  {rows:,} rows × {cols} columns")
                
                if i == 0:
                    print(f"Columns: {list(file_df.columns)}")
                    print("Sample data:")
                    print(file_df.head(2))
                
                successful_dfs.append(file_df)
                
            except Exception as e:
                print(f"✗ Error reading file: {str(e)}")
                print("  This file appears to be corrupted or invalid")
        
        # Combine successful dataframes if any were read
        if successful_dfs:
            print("\nCombining successfully read files...")
            price_combined_df = pd.concat(successful_dfs, ignore_index=True)
            print(f"Combined dataset: {len(price_combined_df):,} rows")
            return price_combined_df  # ✅ Returning the correct variable

        else:
            print("\nNo files could be read successfully")
            return None
            
    except Exception as e:
        print(f"Overall error: {str(e)}")
        return None

# ✅ Assign the returned dataframe to `price_combined_df`
if __name__ == "__main__":
    price_combined_df = test_individual_file_reading()

# ✅ Ensure `price_combined_df` is not None before calling `.head()`
if price_combined_df is not None:
    print(price_combined_df.head(50))
else:
    print("No data to display")
```
