```python
import pandas as pd

# Define the file path
file_path = '/Volumes/T7/MMSDM_2023_12/MMSDM_Historical_Data_SQLLoader/DATA/PUBLIC_DVD_TRADINGPRICE_202312010000.CSV'

# Read the CSV file into a DataFrame, skipping the first row
df = pd.read_csv(file_path, skiprows=1)

# Display general information about the DataFrame
print("DataFrame Info:")
df.info()

# Show the first few rows of the DataFrame
print("\nDataFrame Head:")
print(df.head())

# Show the last few rows of the DataFrame
print("\nDataFrame Tail:")
print(df.tail())

# Display descriptive statistics for numerical columns
print("\nDataFrame Description:")
print(df.describe(include='all'))

# List all column names
print("\nDataFrame Columns:")
print(df.columns.tolist())
```
