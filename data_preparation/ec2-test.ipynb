```python
import pandas as pd
import numpy as np
import awswrangler as wr
import time

# Add timing and logging
print("Starting data load...")
start_time = time.time()

# # Read all Parquet files from the S3 folder
# s3_path = "s3://thesis--ec331-s3/FCAS_RAISE1SEC-Volume-Bids/PUBLIC_DVD_BIDPEROFFER1_202311010000.parquet/"
# df = wr.s3.read_parquet(path=s3_path)

print(f"Data loaded in {time.time() - start_time:.2f} seconds")
print(f"DataFrame shape: {df.shape}")
print(f"Memory usage: {df.memory_usage().sum() / 1024 / 1024:.2f} MB")

# Print some sample data to understand structure
print("\nSample data:")
sample = filtered_df.sample(3)
print(sample[["DUID", "TRADINGDATE", "PERIODID", "MAXAVAIL"] + 
      [f"PRICEBAND{i}" for i in range(1, 11)] + 
      [f"BANDAVAIL{i}" for i in range(1, 11)]])

# Define a more vectorized approach for capping
def cap_bands_vectorized(df):
    print(f"Processing {len(df)} rows with vectorized approach...")
    start_time = time.time()
    
    # Create a copy to avoid modifying the original
    result_df = df.copy()
    
    # Process each row
    for idx, row in df.iterrows():
        max_avail = row["MAXAVAIL"]
        
        # Extract prices and volumes
        prices = [row.get(f"PRICEBAND{i}", np.nan) for i in range(1, 11)]
        volumes = [row.get(f"BANDAVAIL{i}", 0.0) for i in range(1, 11)]
        
        # Replace NaN volumes with 0
        volumes = [0.0 if pd.isna(v) else v for v in volumes]
        
        # Create a list of (price, volume, band_index)
        bands = [(p, v, i) for i, (p, v) in enumerate(zip(prices, volumes), 1) if not pd.isna(p)]
        
        # Sort by price
        bands.sort(key=lambda x: x[0])
        
        # Apply capping
        running_sum = 0.0
        capped_volumes = [0.0] * 10
        
        for price, volume, idx in bands:
            if running_sum >= max_avail:
                break
                
            remaining = max_avail - running_sum
            capped_volume = min(volume, remaining)
            
            capped_volumes[idx-1] = capped_volume
            running_sum += capped_volume
        
        # Update the result dataframe
        for i in range(1, 11):
            col = f"BANDAVAIL{i}"
            if col in result_df.columns:
                result_df.at[idx, col] = capped_volumes[i-1]
    
    print(f"Vectorized processing completed in {time.time() - start_time:.2f} seconds")
    return result_df

# Process in smaller chunks
print("\nProcessing data in chunks...")
chunk_size = 500
capped_dfs = []

for i in range(0, len(df), chunk_size):
    print(f"Processing chunk {i//chunk_size + 1}/{(len(df) + chunk_size - 1)//chunk_size}")
    chunk_start = time.time()
    
    # Get chunk
    chunk = df.iloc[i:i+chunk_size].copy()
    
    # Process chunk
    capped_chunk = chunk.groupby(["DUID", "TRADINGDATE", "PERIODID"]).apply(cap_bands_vectorized)
    capped_dfs.append(capped_chunk)
    
    print(f"Chunk completed in {time.time() - chunk_start:.2f} seconds")

# Combine results
if capped_dfs:
    capped_df = pd.concat(capped_dfs)
    print(f"\nAll processing completed in {time.time() - start_time:.2f} seconds")
    print(capped_df.head())
else:
    print("No results produced")
```
