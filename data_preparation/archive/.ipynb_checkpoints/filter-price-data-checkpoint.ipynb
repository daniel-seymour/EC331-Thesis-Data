```python
# Filter the price bid data 
# Create the output directory if it doesn't already exist
output_dir = '/Volumes/T7/bid-price-data-filtered'
os.makedirs(output_dir, exist_ok=True)

# List all Feather files in the directory
file_list = glob.glob('/Volumes/T7/bid-price-data-2/*.feather')
file_list.sort()

for file in file_list:
    print(f"Processing {file}...")
    # Read the Feather file
    df = pd.read_feather(file)
    
    # Filter to keep rows where BIDTYPE is either "RAISEREG" or "LOWERREG"
    filtered_price_bids = df[(df["BIDTYPE"] == "RAISEREG") | (df["BIDTYPE"] == "LOWERREG")]

    # Print the first few rows of the filtered DataFrame
    print("Filtered DataFrame head:")
    print(filtered_price_bids.head())

    # Construct a new output file path by replacing the directory and file extension
    out_file = file.replace('bid-price-data', 'bid-price-filtered').replace('.feather', '.parquet')

    # Save the filtered DataFrame as a Parquet file
    filtered_price_bids.to_parquet(out_file, index=False)
    print(f"Finished processing {file}.\n")

    # Get just the filename without the path
    base_filename = os.path.basename(file)  # e.g. "PUBLIC_DVD_BIDPEROFFER_D_201201010000.feather"
    # Remove '.feather' extension
    filename_no_ext = os.path.splitext(base_filename)[0]  # e.g. "PUBLIC_DVD_BIDPEROFFER_D_201201010000"

    # Construct the full output path in output_dir
    out_file = os.path.join(output_dir, filename_no_ext + ".parquet")

    # Save to Parquet
    filtered_price_bids.to_parquet(out_file, index=False)
    print(f"Finished processing {file} -> {out_file}\n")
```

```python
import os
import glob
import pandas as pd
import pyarrow.feather as feather  # for reading Feather files
from tqdm import tqdm

# Define input/output directories
input_dir = '/Volumes/T7/bid-price-data-sorted-feather'
output_dir = '/Volumes/T7/bid-price-data-excluding-energy'
os.makedirs(output_dir, exist_ok=True)

# Gather all Feather files
feather_files = sorted(glob.glob(os.path.join(input_dir, '*.feather')))

# Initialize progress bar
with tqdm(total=len(feather_files), desc="Processing files") as pbar:
    for file_path in feather_files:
        try:
            # Read Feather file into a pandas DataFrame
            table = feather.read_table(file_path)  # returns a pyarrow.Table
            df = table.to_pandas()                 # convert to pandas DataFrame

            # Filter out rows where BIDTYPE is 'ENERGY'
            filtered_df = df[df['BIDTYPE'] != 'ENERGY']
            num_filtered = len(filtered_df)
            print(f"{os.path.basename(file_path)} -> {num_filtered} rows after filtering.")

            if num_filtered > 0:
                # Construct output file path (replacing '.feather' with '.parquet')
                base_filename = os.path.basename(file_path).replace('.feather', '.parquet')
                out_path = os.path.join(output_dir, base_filename)

                # Write filtered data to Parquet
                filtered_df.to_parquet(out_path, index=False)
                print(f"Filtered data saved to {out_path}\n")
            else:
                print("No data left after filtering, skipping write.\n")

        except Exception as e:
            print(f"Error processing {file_path}: {e}\n")

        pbar.update(1)

print("All files processed!")
```
